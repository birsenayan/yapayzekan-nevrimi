## B. Başlangıç: veriyi oluşturma ve okuma

Aşağıdaki hücre, örnek veri tablosunu oluşturur ve CSV dosyasına kaydeder. Kendi dosyanı kullanmak istersen `veri_yolu` değişkenini dosyanın adıyla değiştirebilirsin.

```python
import pandas as pd
from pathlib import Path

kayitlar = [
    {"yil": 1950, "gelisme": "Turing'in makine zekâsı makalesi", "alan": "Kuramsal temel"},
    {"yil": 1956, "gelisme": "Dartmouth yaz araştırma projesi", "alan": "Araştırma"},
    {"yil": 1966, "gelisme": "ELIZA sohbet programı", "alan": "Doğal dil"},
    {"yil": 1972, "gelisme": "MYCIN uzman sistemi", "alan": "Uzman sistemler"},
    {"yil": 1980, "gelisme": "Uzman sistemlerin yaygınlaşması", "alan": "Uzman sistemler"},
    {"yil": 1997, "gelisme": "Deep Blue'nun Kasparov'u yenmesi", "alan": "Oyun ve arama"},
    {"yil": 2006, "gelisme": "Derin öğrenme araştırmalarında ivme", "alan": "Makine öğrenmesi"},
    {"yil": 2011, "gelisme": "IBM Watson'ın Jeopardy! başarısı", "alan": "Doğal dil"},
    {"yil": 2012, "gelisme": "AlexNet'in ImageNet başarısı", "alan": "Derin öğrenme"},
    {"yil": 2016, "gelisme": "AlphaGo'nun Lee Sedol'u yenmesi", "alan": "Derin öğrenme"},
    {"yil": 2017, "gelisme": "Transformer mimarisinin tanıtılması", "alan": "Dil modelleri"},
    {"yil": 2020, "gelisme": "GPT-3'ün yayımlanması", "alan": "Üretken yapay zekâ"},
    {"yil": 2022, "gelisme": "ChatGPT'nin kullanıma sunulması", "alan": "Üretken yapay zekâ"},
    {"yil": 2023, "gelisme": "GPT-4'ün tanıtılması", "alan": "Çok modlu yapay zekâ"},
    {"yil": 2024, "gelisme": "Çok modlu modellerde ilerlemeler", "alan": "Çok modlu yapay zekâ"},
    {"yil": 2025, "gelisme": "Akıl yürütme ve araç kullanımı odaklı sistemlerin ilerlemesi", "alan": "Akıl yürütme ve ajanlar"}
]

veri = pd.DataFrame(kayitlar)
veri.to_csv("yapay_zeka_tarihcesi.csv", index=False, encoding="utf-8-sig")

# CSV dosyasını yeniden okuyarak dosya okuma işlemini gösteriyoruz.
veri_yolu = "yapay_zeka_tarihcesi.csv"
veri = pd.read_csv(veri_yolu)
print(veri.head())
```

## C. Veri inceleme ve hesaplama

Bu bölümde veri setinin boyutunu, eksik değerlerini, yıl aralığını ve alanlara göre dağılımını inceliyoruz.

```python
print("Veri boyutu (satır, sütun):", veri.shape)
print("\\nEksik değer sayıları:")
print(veri.isna().sum())
print("\\nİlk yıl:", veri["yil"].min())
print("Son yıl:", veri["yil"].max())
print("Toplam seçilmiş gelişme:", len(veri))
print("\\nAlanlara göre gelişme sayısı:")
print(veri["alan"].value_counts())
```

## D. Grafik ve sonuç dosyaları

İlk grafik, seçilmiş gelişmelerin on yıllara göre kaç tane olduğunu gösterir. İkinci grafik ise veri setindeki seçilmiş gelişmelerin birikimli sayısını gösterir. Bu grafikler tarihsel ilerlemenin eksiksiz ölçümü değildir; yalnızca oluşturduğumuz örnek veri setini görselleştirir.

```python
import matplotlib.pyplot as plt
from pathlib import Path

sonuc_klasoru = Path("sonuclar")
sonuc_klasoru.mkdir(exist_ok=True)

veri["on_yil"] = (veri["yil"] // 10) * 10
on_yil_sayisi = veri.groupby("on_yil").size().sort_index()

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar([str(y) + "'lar" for y in on_yil_sayisi.index], on_yil_sayisi.values)
ax.set_title("Seçilen yapay zekâ gelişmelerinin on yıllara göre dağılımı")
ax.set_xlabel("On yıl")
ax.set_ylabel("Seçilmiş gelişme sayısı")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
fig.savefig(sonuc_klasoru / "on_yillara_gore_gelismeler.png", dpi=160)
plt.show()
```

```python
yil_sayisi = veri.groupby("yil").size().sort_index()
birikimli = yil_sayisi.cumsum()

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(birikimli.index, birikimli.values, marker="o")
ax.set_title("Seçilmiş gelişmelerin yıllara göre birikimli sayısı")
ax.set_xlabel("Yıl")
ax.set_ylabel("Birikimli gelişme sayısı")
ax.grid(alpha=0.25)
fig.tight_layout()
fig.savefig(sonuc_klasoru / "birikimli_gelisim.png", dpi=160)
plt.show()
```

## F. CSV, JSON ve PNG olarak kaydetme

Aşağıdaki kod, veri tablosunu CSV ve JSON; grafikleri ise PNG olarak kaydeder.

```python
veri.to_csv(sonuc_klasoru / "yapay_zeka_tarihcesi.csv", index=False, encoding="utf-8-sig")
veri.to_json(sonuc_klasoru / "yapay_zeka_tarihcesi.json", orient="records", force_ascii=False, indent=2)

print("Kaydedilen dosyalar:")
for dosya in sorted(sonuc_klasoru.iterdir()):
    print("-", dosya.name)
```
